```python
import json
import os
import re
import urllib.request

sample_notes = """Website revamp kickoff
Alex: What are the main features you want on the new landing page
Client: We need a modern hero section, customer testimonials, and
Alex: Would you prefer Calendly integration or a custom-built sol
Client: Calendly works fine and keeps it simple.
Alex: Perfect, I ll prepare the initial wireframes by Wednesday."""

def offline_extract(notes):
    """Extract lines in the format Name: task by deadline."""
    actions = []
    # Matches "Name: ... by Deadline" at the end of a line
    pattern = re.compile(r"^([A-Za-z]+):\s*(.+?)\s+by\s+([A-Za-z0-9\s]+)\.?$", re.IGNORECASE)
    
    for line in notes.strip().splitlines():
        match = pattern.match(line.strip())
        if match:
            actions.append({
                "assignee": match.group(1),
                "task": match.group(2).strip(),
                "deadline": match.group(3).strip()
            })
    return actions

if __name__ == "__main__":
    extracted_actions = offline_extract(sample_notes)
    print(json.dumps(extracted_actions, indent=2))
```

```python
# Install required library
%pip install ipywidgets

# Import libraries
import ipywidgets as widgets
from IPython.display import display

# Sample notes
example_notes = """
Python is a popular programming language.
Python is used in AI, machine learning, web development and data analysis.
Python is easy to learn for beginners.
Python supports libraries like pandas, numpy and matplotlib.
Python can be used to automate repetitive tasks.
"""

# Create notes text box
notes_box = widgets.Textarea(
    value=example_notes,
    description="Notes:",
    layout=widgets.Layout(width="100%", height="180px")
)

# Create mode selection
mode_box = widgets.ToggleButtons(
    options=["Original Notes", "Simple Summary", "Important Points"],
    value="Original Notes",
    description="Mode:"
)

# Create button
button = widgets.Button(
    description="Process Notes",
    button_style="success",
    icon="check"
)

# Create output area
output = widgets.Output()

# Create function
def process_notes(b):
    with output:
        output.clear_output()

        notes = notes_box.value.strip()
        mode = mode_box.value

        if not notes:
            print("Please enter your notes first.")
            return

        if mode == "Original Notes":
            print("Original Notes:\n")
            print(notes)

        elif mode == "Simple Summary":
            sentences = notes.replace("\n", " ").split(".")
            sentences = [s.strip() for s in sentences if s.strip()]
            summary = ". ".join(sentences[:3])

            print("Simple Summary:\n")
            print(summary + "." if summary else "No notes available.")

        elif mode == "Important Points":
            sentences = notes.replace("\n", " ").split(".")
            sentences = [s.strip() for s in sentences if s.strip()]

            print("Important Points:\n")

            for i, sentence in enumerate(sentences, start=1):
                print(f"{i}. {sentence}")

# Connect button
button.on_click(process_notes)

# Display application
display(notes_box, mode_box, button, output)
```
