```python

# ================================================
# SMART RAG CHATBOT - BROWSER BASED DEMO
# ================================================

import sys
import subprocess

# Install Gradio only if it is not available
try:
    import gradio as gr
except ImportError:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "-q", "gradio"]
    )
    import gradio as gr

# Retrieval library
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample knowledge base
documents = [
    """Artificial Intelligence is a technology that enables machines
    to perform tasks that usually require human intelligence.
    Machine learning allows computers to learn from data.""",

    """Retrieval Augmented Generation is called RAG.
    RAG retrieves relevant information from a knowledge base
    and uses the retrieved information to generate answers.""",

    """Ollama is a platform that allows developers to run
    open-source language models locally on their computers.""",

    """Python is a popular programming language used for AI,
    machine learning, automation, web development and data analysis.""",

    """Vector search helps identify relevant information
    by comparing numerical representations of text.""",

    """A chatbot is an application that communicates with users
    and answers their questions using programmed logic or AI models.""",

    """RAG has three main stages: document processing,
    information retrieval and answer generation.""",

    """Machine learning is a subset of artificial intelligence
    that enables computers to learn patterns from data.""",

    """Deep learning uses neural networks to learn complex patterns
    from large amounts of data.""",

    """Natural Language Processing enables computers to understand,
    process and interact with human language."""
]

# Create text retrieval system
vectorizer = TfidfVectorizer(stop_words="english")
document_vectors = vectorizer.fit_transform(documents)

# Retrieval function
def retrieve_information(question):
    query_vector = vectorizer.transform([question])

    scores = cosine_similarity(
        query_vector, document_vectors
    ).flatten()

    best_indices = scores.argsort()[-3:][::-1]

    context = "\n\n".join(
        documents[i] for i in best_indices if scores[i] > 0
    )

    return context

# Chatbot function
def chatbot(question):
    if not question.strip():
        return "Please enter a question."

    context = retrieve_information(question)

    if not context:
        return "I couldn't find relevant information in my knowledge base."

    return (
        "ANSWER BASED ON MY KNOWLEDGE BASE:\n\n"
        + context
        + "\n\n---\n"
        + "Retrieved using TF-IDF similarity search."
    )

# Create interface
demo = gr.Interface(
    fn=chatbot,
    inputs=gr.Textbox(
        label="Ask a question",
        placeholder="What is RAG?"
    ),
    outputs=gr.Textbox(
        label="Chatbot Answer",
        lines=10
    ),
    title="Smart RAG Document Chatbot",
    description=(
        "A simple Retrieval-Augmented Generation prototype. "
        "Ask questions about AI, RAG, Python, Ollama and NLP."
    ),
    examples=[
        "What is RAG?",
        "What is artificial intelligence?",
        "What is Ollama?",
        "What is deep learning?",
        "What is Python?"
    ],
    theme="soft"
)

# Launch chatbot
demo.launch(share=True)
```
